```python
def make_rectangle(points):
    min_x = min(x for x, y in points)
    max_x = max(x for x, y in points)

    min_y = min(y for x, y in points)
    max_y = max(y for x, y in points)

    rectangle = [
        (min_x, min_y),
        (max_x, min_y),
        (max_x, max_y),
        (min_x, max_y)
    ]

    return rectangle


points = [(2, 3), (7, 5), (4, 9), (1, 6)]

rectangle = make_rectangle(points)

print(rectangle)
```

```python
import matplotlib.pyplot as plt

points = [(2, 3), (7, 5), (4, 9), (1, 6)]

rectangle = make_rectangle(points)

# Close the rectangle by repeating the first point
x = [p[0] for p in rectangle] + [rectangle[0][0]]
y = [p[1] for p in rectangle] + [rectangle[0][1]]

plt.plot(x, y, marker="o")
plt.scatter(*zip(*points))

plt.xlabel("x")
plt.ylabel("y")
plt.grid()
plt.axis("equal")
plt.show()
```
